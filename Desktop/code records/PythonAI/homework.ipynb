# Python tapşırıqları

Temperatur, gündəlik xərclər, meyvə səsləri və fənn balları üzrə hesablamalar.

## Tapşırıq 1 — Temperatur göstəriciləri

```python
temperatures = [2, 5, 4, 6, 7, 3, 1]

print("Orta qiymət:", sum(temperatures) / len(temperatures))
print("Maksimum:", max(temperatures))
print("Minimum:", min(temperatures))
print("Fərq:", max(temperatures) - min(temperatures))
```

## Tapşırıq 2 — Gündəlik xərclər

```python
expenses = [1200, 1500, 800, 600, 1000, 400, 900]

print("Ümumi xərc:", sum(expenses))
print("Orta xərc:", sum(expenses) / len(expenses))
print("Ən yüksək xərc:", max(expenses))
print("Ən aşağı xərc:", min(expenses))
print("1000-dən çox xərc olan günlər:", sum(expense > 1000 for expense in expenses))
```

## Tapşırıq 3 — Meyvələr üzrə səslər

```python
votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "orange"]

counts = {fruit: votes.count(fruit) for fruit in votes}

for fruit, count in counts.items():
    print(fruit, "→", count)

leader = max(counts, key=counts.get)
print("Lider:", leader)

def get_count(item):
    return item[1]

sorted_votes = sorted(counts.items(), key=get_count, reverse=True)
print("İlk 3:", sorted_votes[:3])
```

## Tapşırıq 4 — Fənlər üzrə nəticələr

```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {"Math": math, "Physics": physics, "English": english}

for subject, scores in subjects.items():
    print(subject, "orta:", sum(scores) / len(scores), "minimum:", min(scores), "maksimum:", max(scores))

def get_average(subject):
    return sum(subjects[subject]) / len(subjects[subject])

best_subject = max(subjects, key=get_average)
best_subject
```
